```python
# Bài 7.1: Tạo từ điển kích thước N với giá trị x^3
def cube_dict(n):
    return {x: x**3 for x in range(1, n+1)}

print(cube_dict(5))
```

```python
# Bài 7.2: Lưu trữ tên và điểm thi, ánh xạ điểm sang xếp loại A-F
def grade_from_score(score):
    if score >= 90:
        return 'A'
    if score >= 80:
        return 'B'
    if score >= 70:
        return 'C'
    if score >= 60:
        return 'D'
    return 'F'

students = {'An': 95, 'Binh': 82, 'Chi': 68, 'Duc': 54}
grades = {name: grade_from_score(score) for name, score in students.items()}
print(grades)
```

```python
# Bài 7.3: Đếm số lượng sinh viên theo mức xếp loại
def count_grade_levels(grades):
    counts = {}
    for value in grades.values():
        counts[value] = counts.get(value, 0) + 1
    return counts

print(count_grade_levels({'An': 'A', 'Binh': 'B', 'Chi': 'D', 'Duc': 'F'}))
```

```python
# Bài 7.4: Làm sạch văn bản tiếng Anh và đếm từ
import re

def word_frequency(text):
    cleaned = re.sub(r'[^A-Za-z0-9\s]', '', text).lower()
    words = cleaned.split()
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    return counts

print(word_frequency('Hello, hello world! This is a test. Hello world.'))
```

```python
# Bài 7.5: Tìm từ có tần suất cao nhất và thấp nhất
def most_least_common(words):
    if not words:
        return None, None
    most = max(words, key=words.get)
    least = min(words, key=words.get)
    return (most, words[most]), (least, words[least])

freq = {'hello': 3, 'world': 2, 'test': 1}
print(most_least_common(freq))
```

```python
# Bài 7.6: Thêm khóa pocket và cập nhật gold trong inventory
def update_inventory(inv):
    inv['pocket'] = inv.get('pocket', []) + ['map', 'torch']
    inv['gold'] = inv.get('gold', 0) + 50
    return inv

inventory = {'backpack': ['rope', 'knife'], 'gold': 100}
print(update_inventory(inventory))
```

```python
# Bài 7.7: Sắp xếp đồ vật trong backpack và loại bỏ một vật phẩm
def sort_and_remove_item(inv, item):
    items = sorted(inv.get('backpack', []))
    if item in items:
        items.remove(item)
    inv['backpack'] = items
    return inv

inventory = {'backpack': ['apple', 'bread', 'coin']}
print(sort_and_remove_item(inventory, 'bread'))
```

```python
# Bài 7.8: Tính hóa đơn chi tiết từ số lượng và đơn giá
def invoice(items, prices):
    total = 0
    invoice_lines = []
    for item, qty in items.items():
        cost = qty * prices.get(item, 0)
        invoice_lines.append((item, qty, prices.get(item, 0), cost))
        total += cost
    return invoice_lines, total

print(invoice({'apple': 2, 'milk': 1}, {'apple': 3, 'milk': 5}))
```

```python
# Bài 7.9: Khấu trừ số lượng vật phẩm đã giao và cập nhật inventory
def deduct_inventory(inv, used):
    for item, qty in used.items():
        if item in inv:
            inv[item] = max(inv[item] - qty, 0)
    return inv

inventory = {'apple': 10, 'bread': 5}
print(deduct_inventory(inventory, {'apple': 3, 'bread': 2}))
```

```python
# Bài 7.10: Dùng set để tìm các sản phẩm có trong kho nhưng chưa được khách chọn
def available_not_chosen(stock, chosen):
    return set(stock) - set(chosen)

print(available_not_chosen(['apple', 'banana', 'cherry'], ['banana']))
```
