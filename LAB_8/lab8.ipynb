```python
# Bài 8.1: Hàm kiểm tra số nguyên tố và các cặp số nguyên tố sinh đôi < 1000
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

twins = [(i, i+2) for i in range(2, 1000) if is_prime(i) and is_prime(i+2)]
print(twins[:10])
```

```python
# Bài 8.2: Hàm tính giai thừa
def factorial(n):
    result = 1
    for i in range(2, n+1):
        result *= i
    return result

print(factorial(6))
```

```python
# Bài 8.3: Hàm hoán vị và tổ hợp
def permutation(n, r):
    return factorial(n) // factorial(n-r) if 0 <= r <= n else 0

def combination(n, r):
    return permutation(n, r) // factorial(r) if 0 <= r <= n else 0

print(permutation(5, 3), combination(5, 3))
```

```python
# Bài 8.4: Hàm cubesum tính tổng lập phương chữ số
def cubesum(n):
    return sum(int(d)**3 for d in str(abs(n)))

print(cubesum(153))
```

```python
# Bài 8.5: Hàm isArmstrong và danh sách số Armstrong
def is_armstrong(n):
    return n == cubesum(n)

armstrong_numbers = [x for x in range(1, 1000) if is_armstrong(x)]
print(armstrong_numbers)
```

```python
# Bài 8.6: Hàm sumPdivisors tính tổng ước số thực sự
def sum_proper_divisors(n):
    if n <= 1:
        return 0
    return sum(i for i in range(1, n//2 + 1) if n % i == 0)

print(sum_proper_divisors(28))
```

```python
# Bài 8.7: Kiểm tra cặp số amicable
def are_amicable(a, b):
    return a != b and sum_proper_divisors(a) == b and sum_proper_divisors(b) == a

print(are_amicable(220, 284))
```

```python
# Bài 8.8: Dùng filter và lambda tách số chẵn, lẻ
def split_even_odd(numbers):
    evens = list(filter(lambda x: x % 2 == 0, numbers))
    odds = list(filter(lambda x: x % 2 != 0, numbers))
    return evens, odds

print(split_even_odd([1,2,3,4,5,6]))
```

```python
# Bài 8.9: Dùng map để tính lập phương cho mỗi phần tử
def cubes(numbers):
    return list(map(lambda x: x**3, numbers))

print(cubes([1,2,3,4]))
```

```python
# Bài 8.10: Kết hợp filter và map cho số chẵn và lẻ
def even_cubes_odd_squares(numbers):
    evens = list(map(lambda x: x**3, filter(lambda x: x % 2 == 0, numbers)))
    odds = list(map(lambda x: x**2, filter(lambda x: x % 2 != 0, numbers)))
    return evens, odds

print(even_cubes_odd_squares([1,2,3,4,5]))
```
