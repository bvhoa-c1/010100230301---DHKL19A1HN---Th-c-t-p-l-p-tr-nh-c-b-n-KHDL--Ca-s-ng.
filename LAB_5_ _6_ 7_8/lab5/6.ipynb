```python
# 1. Khởi tạo một dictionary trống
my_dict = {}

# 2. Thêm hoặc cập nhật một phần tử
my_dict["name"] = "Alice"  # Thêm key "name" với value "Alice"
my_dict["age"] = 20        # Thêm key "age" với value 20

# 3. Truy xuất giá trị thông qua Key
print(my_dict["name"])  # Kết quả: Alice

# 4. Duyệt qua từ điển bằng vòng lặp for
for key, value in my_dict.items():
    print(f"{key}: {value}")
```
