```python
# Bài 6.1: Phân loại và tính tổng số chẵn, số lẻ
def sum_even_odd(numbers):
    even_sum = sum(x for x in numbers if x % 2 == 0)
    odd_sum = sum(x for x in numbers if x % 2 != 0)
    return even_sum, odd_sum

print(sum_even_odd([1, 2, 3, 4, 5, 6]))
```

```python
# Bài 6.2: Tìm số nguyên tố và số hoàn hảo trong mảng
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def is_perfect(n):
    return n > 1 and sum(i for i in range(1, n) if n % i == 0) == n

def prime_or_perfect(numbers):
    return [x for x in numbers if is_prime(x) or is_perfect(x)]

print(prime_or_perfect([2, 3, 6, 28, 10, 11]))
```

```python
# Bài 6.3: Tìm giá trị lớn nhất và nhỏ nhất trong dãy hỗn hợp
def min_max(sequence):
    return min(sequence), max(sequence)

print(min_max([3, 7.5, 2, 9.1, -1, 0]))
```

```python
# Bài 6.4: List comprehension tạo danh sách Fibonacci
def fibonacci_list(n):
    fib = [0, 1]
    while len(fib) < n:
        fib.append(fib[-1] + fib[-2])
    return fib[:n]

print(fibonacci_list(10))
```

```python
# Bài 6.5: List comprehension cho các số nguyên tố < 100
def primes_under_100():
    def is_prime(x):
        if x < 2:
            return False
        for i in range(2, int(x**0.5) + 1):
            if x % i == 0:
                return False
        return True
    return [x for x in range(2, 100) if is_prime(x)]

print(primes_under_100())
```

```python
# Bài 6.6: Kiểm tra cấp số cộng bằng sai phân giữa các phần tử liên tiếp
def is_arithmetic_sequence(numbers):
    if len(numbers) < 2:
        return True
    diffs = [numbers[i+1] - numbers[i] for i in range(len(numbers)-1)]
    return all(d == diffs[0] for d in diffs), diffs

print(is_arithmetic_sequence([2, 5, 8, 11]))
```

```python
# Bài 6.7: Tính tổng phần tử trong ma trận
def matrix_sum(matrix):
    return sum(sum(row) for row in matrix)

print(matrix_sum([[1, 2, 3], [4, 5, 6]]))
```

```python
# Bài 6.8: Nhân hai ma trận nếu điều kiện phù hợp
def multiply_matrices(A, B):
    if not A or not B or len(A[0]) != len(B):
        return None
    m, p, n = len(A), len(B), len(B[0])
    return [[sum(A[i][k] * B[k][j] for k in range(p)) for j in range(n)] for i in range(m)]

print(multiply_matrices([[1,2],[3,4]], [[5,6],[7,8]]))
```

```python
# Bài 6.9: Tạo ma trận chuyển vị và kiểm tra ma trận đối xứng
def transpose(matrix):
    return [list(row) for row in zip(*matrix)]

def is_symmetric(matrix):
    return matrix == transpose(matrix)

mat = [[1,2,3],[2,5,6],[3,6,9]]
print(transpose(mat))
print('symmetric=', is_symmetric(mat))
```

```python
# Bài 6.10: Tìm ma trận nghịch đảo bằng phương pháp Gauss-Jordan
def invert_matrix(matrix):
    n = len(matrix)
    if any(len(row) != n for row in matrix):
        return None
    aug = [row[:] + [1 if i == j else 0 for j in range(n)] for i, row in enumerate(matrix)]
    for i in range(n):
        pivot = aug[i][i]
        if pivot == 0:
            for j in range(i+1, n):
                if aug[j][i] != 0:
                    aug[i], aug[j] = aug[j], aug[i]
                    pivot = aug[i][i]
                    break
        if pivot == 0:
            return None
        aug[i] = [x / pivot for x in aug[i]]
        for j in range(n):
            if j != i:
                factor = aug[j][i]
                aug[j] = [aug[j][k] - factor * aug[i][k] for k in range(2*n)]
    return [row[n:] for row in aug]

print(invert_matrix([[4,7],[2,6]]))
```
