```python
# Bài 5.1: Chuyển đổi số nguyên dương thập phân sang nhị phân
def decimal_to_binary(n):
    return bin(n)[2:]

print(decimal_to_binary(45))  # Ví dụ: 45 -> 101101
```

```python
# Bài 5.2: Tìm chuỗi con chung có độ dài nhỏ nhất giữa hai chuỗi
def shortest_common_substrings(a, b):
    common = set()
    for i in range(len(a)):
        for j in range(i + 1, len(a) + 1):
            sub = a[i:j]
            if sub and sub in b:
                common.add(sub)
    if not common:
        return []
    min_len = min(len(x) for x in common)
    return [x for x in common if len(x) == min_len]  # Các chuỗi ngắn nhất

print(shortest_common_substrings('abcde', 'xbcz'))
```

```python
# Bài 5.3: Tìm vị trí xuất hiện của từ khóa và từ xuất hiện nhiều nhất
import re

def keyword_positions_and_top_word(text, keyword):
    positions = [m.start() for m in re.finditer(re.escape(keyword), text, re.IGNORECASE)]
    words = re.findall(r'\b\w+\b', text.lower())
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    top_word = max(freq, key=freq.get) if freq else None
    return positions, top_word, freq.get(top_word, 0)

positions, top_word, count = keyword_positions_and_top_word('Hello world. The world is wide.', 'world')
print('positions=', positions, 'top_word=', top_word, 'count=', count)
```

```python
# Bài 5.4: Loại bỏ ký tự không phải chữ số, chuyển thành số nguyên và kiểm tra nguyên tố
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def clean_and_check_prime(text):
    digits = ''.join(ch for ch in text if ch.isdigit())
    if not digits:
        return None, False
    n = int(digits)
    return n, is_prime(n)

print(clean_and_check_prime('a12b3c45'))
```

```python
# Bài 5.5: Trộn hai chuỗi theo kiểu xen kẽ và phân tách bằng dấu gạch nối
def interleave_strings(str1, str2):
    result = []
    for a, b in zip(str1, str2):
        result.append(a)
        result.append(b)
    result.extend(str1[len(str2):] or str2[len(str1):])
    return '-'.join(''.join(result)[i:i+1] for i in range(len(''.join(result))))

print(interleave_strings('abc', 'XYZ'))
```

```python
# Bài 5.6: Đếm ký tự đặc biệt và tỷ lệ phần trăm của chúng
def special_char_stats(text):
    special = [ch for ch in text if not ch.isalnum()]
    total = len(text)
    counts = {}
    for ch in special:
        counts[ch] = counts.get(ch, 0) + 1
    return {ch: (count, round(count / total * 100, 2)) for ch, count in counts.items()}

print(special_char_stats('Hello! How are you? #python @2026.'))
```

```python
# Bài 5.7: Thống kê chữ thường, chữ hoa, chữ số và ký tự đặc biệt
def character_breakdown(text):
    stats = {'lower': 0, 'upper': 0, 'digit': 0, 'special': 0}
    for ch in text:
        if ch.islower():
            stats['lower'] += 1
        elif ch.isupper():
            stats['upper'] += 1
        elif ch.isdigit():
            stats['digit'] += 1
        else:
            stats['special'] += 1
    return stats

print(character_breakdown('Hello World! 123'))
```

```python
# Bài 5.8: Trích xuất chuỗi con và biến đổi chữ hoa/chữ thường
def substring_transform(text):
    if len(text) <= 10:
        return None
    part_2_8 = text[1:8]  # vị trí 2 đến 8 nếu lấy theo chỉ số 1-based
    part_5_9 = text[4:9]  # 5 ký tự từ vị trí 5
    last_3 = text[-3:]
    upper = text.upper()
    lower = text.lower()
    reverse = text[::-1]
    return part_2_8, part_5_9, last_3, upper, lower, reverse

print(substring_transform('HelloPythonWorld'))
```

```python
# Bài 5.9: Đánh giá khả năng chuyển đổi chuỗi bằng thêm, xóa hoặc thay thế ký tự
def edit_distance(a, b):
    dp = [[0] * (len(b) + 1) for _ in range(len(a) + 1)]
    for i in range(len(a) + 1):
        dp[i][0] = i
    for j in range(len(b) + 1):
        dp[0][j] = j
    for i in range(1, len(a) + 1):
        for j in range(1, len(b) + 1):
            if a[i-1] == b[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = 1 + min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1])
    return dp[-1][-1]

print(edit_distance('kitten', 'sitting'))
```

```python
# Bài 5.10: Loại bỏ toàn bộ ký tự khoảng trắng trong xâu
def remove_whitespace(text):
    return ''.join(ch for ch in text if not ch.isspace())

print(remove_whitespace('Hello   World 	 2026'))
```
